```python
from pathlib import Path
from PIL import Image
import pandas as pd

image_root = Path("/content/pilot_images")

unexpected = []

for image_path in image_root.rglob("*.jpg"):
    with Image.open(image_path) as image:
        if image.format != "JPEG":
            unexpected.append(
                {
                    "path": str(image_path),
                    "format": image.format,
                    "width": image.width,
                    "height": image.height,
                }
            )

unexpected_df = pd.DataFrame(unexpected)

print(unexpected_df.to_string(index=False))
```

```python
import hashlib
from collections import defaultdict
from pathlib import Path

image_root = Path("/content/pilot_images")

hash_to_paths = defaultdict(list)

for image_path in image_root.rglob("*"):
    if not image_path.is_file():
        continue

    digest = hashlib.sha256()

    with image_path.open("rb") as file:
        while chunk := file.read(1024 * 1024):
            digest.update(chunk)

    hash_to_paths[digest.hexdigest()].append(image_path)

for digest, paths in hash_to_paths.items():
    if len(paths) > 1:
        print("Duplicate hash:", digest)

        for path in paths:
            print(" ", path)
```

```python
import pandas as pd

evidence = pd.read_parquet(
    "data/manifests/pilot_evidence.parquet"
)

queries = pd.read_parquet(
    "data/manifests/pilot_queries.parquet"
)

duplicate_paths = {
    "visual_news/origin/usa_today/images/0316/125.jpg",
    "visual_news/origin/usa_today/images/0627/512.jpg",
}

duplicate_evidence = evidence[
    evidence["image_path"].isin(duplicate_paths)
]

print("Evidence rows:")
print(
    duplicate_evidence[
        [
            "id",
            "image_path",
            "caption",
            "source",
            "is_query_provenance",
        ]
    ].to_string(index=False)
)

query_matches = queries[
    queries["query_image_path"].isin(duplicate_paths)
]

print("\nQuery rows:")
print(
    query_matches[
        [
            "query_id",
            "id",
            "image_id",
            "falsified",
            "query_image_path",
        ]
    ].to_string(index=False)
)
```
